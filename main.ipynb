Комплексное сопряжение и максимально вещественное подполе

```python
c = K.complex_conjugation()
c(a)
```

```python
for _ in range(5):
    y = K.random_element()
    z = K.random_element()
    #сопряжение - инволюция
    if not c(c(y)) == y:
        print (y)
    #сопряжение сохраняет сложение
    if not c(y+z) == c(y)+c(z):
        print(y, z)
    #сопряжение сохраняет умножение
    if not c(y*z)==c(y)*c(z):
        print(y, z)

#сопряжение не тождественное
if c(a)==a:
    print("плохо плохо неважно")

#перестановочно ли сопряжение с другими автоморфизмами поля
y = K.random_element()
for s in K.automorphisms():
    if c(s(y)) != s(c(y)):
        print(s)
```

```python
#Максимальное вещественное подполе K+ и его вложение в К
Kplus, emb = K.maximal_totally_real_subfield()
print(Kplus)
print("степень К+:", Kplus.degree())
print("образ образующей К+ в К:", emb(Kplus.gen()))
```
