Арифметика в целочисленном базисе

```python
n = K.degree()

#координаты базисных элементов В по строкам в базисе 1, a^1,...

Bmat = matrix(QQ, [b.vector() for b in B])

def to_cords(y):
    #разложение элемента поля по координатам в базисе В
    return Bmat.solve_left(y.vector())

def from_cords(v):
    #преобразование координат в базисе В в элемент поля
    return sum(v[i]*B[i] for i in range(n))

#проверка корректности преобразований
y = OK.random_element()
print(to_cords(y))
print(from_cords(to_cords(y))== y)
```

```python
#таблица умножения - координаты произведения B[i]*B[j]
Mul = [[to_cords(B[i]*B[j]) for j in range(n)]for i in range(n)]

#матрица сопряжения: j-я строка - координаты c(B[j])
C = matrix(QQ, [to_cords(c(B[j])) for j in range(n)])
```

```python
#реализация операций

def add_c(u, v):
    return u+v

def sub_c(u, v):
    return u-v

def mul_c(u, v):
    m = vector(QQ, n) 
    for i in range(n):
        for j in range(n):
            m += u[i]*v[j]*Mul[i][j] #по билинейности умножения
    return m

#матрица умножения на элемент
def mul_el(u):
    #j-я строка - координаты элемента u*B[j]
    rows = [sum((u[i]*Mul[i][j] for i in range(n)), vector(QQ, n)) for j in range(n)]
    return matrix(QQ, rows)

def conj_c(u):
    return u*C

def norm_c(u):
    return mul_el(u).det()

def div_c(u, v):
    return mul_el(v).solve_left(u)

#
for _ in range(5):
    y = OK.random_element()
    z = OK.random_element()
    if z==0: continue

    u, v = to_cords(y), to_cords(z)

    assert from_cords(add_c(u, v)) == y+z
    assert from_cords(sub_c(u, v)) == y-z
    assert from_cords(mul_c(u, v)) == y*z
    assert from_cords(conj_c(u)) == c(y)
    assert norm_c(u) == y.norm()
    assert from_cords(div_c(u, v)) == y/z
print("ок")


```
